```python
import pandas as pd
```

```python
df = pd.read_excel("Dataset for Data Analytics.xlsx")
```

```python
print("Rows and Columns:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
```

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
df.isnull().sum()[df.isnull().sum() > 0]
```

```python
df["CouponCode"] = df["CouponCode"].fillna("NO COUPON")
```

```python
print("Missing CouponCode:", df["CouponCode"].isnull().sum())
```

```python
duplicate_orderids = df["OrderID"].duplicated().sum()

print("Duplicate OrderIDs:", duplicate_orderids)
```

```python
duplicate_rows = df.duplicated().sum()

print("Duplicate complete rows:", duplicate_rows)
```

```python
print(df["Date"].head())
print("\nData type:", df["Date"].dtype)
```

```python
df["Date"] = pd.to_datetime(df["Date"]).dt.strftime("%Y-%m-%d")
```

```python
print(df["Date"].head())
```

```python
print("Quantity:", df["Quantity"].dtype)
print("ItemsInCart:", df["ItemsInCart"].dtype)
print("UnitPrice:", df["UnitPrice"].dtype)
print("TotalPrice:", df["TotalPrice"].dtype)
```

```python
df["Quantity"] = df["Quantity"].round().astype(int)
df["ItemsInCart"] = df["ItemsInCart"].round().astype(int)
```

```python
print(df[["Quantity", "ItemsInCart"]].head())
```

```python
df["UnitPrice"] = df["UnitPrice"].round(2)
df["TotalPrice"] = df["TotalPrice"].round(2)
```

```python
print(df[["UnitPrice", "TotalPrice"]].head())
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("\nBlank values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
df.to_excel("Project1_Cleaned.xlsx", index=False)

print("File saved successfully!")
```
