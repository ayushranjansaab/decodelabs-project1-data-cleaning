```python
import pandas as pd
```

```python
df = pd.read_excel("Dataset for Data Analytics.xlsx")
```

```python
print("Rows and Columns:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
```

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
print("Duplicate Order IDs:", df["OrderID"].duplicated().sum())
```

```python
df.isnull().sum()[df.isnull().sum() > 0]
```

```python
df["CouponCode"] = df["CouponCode"].fillna("NO COUPON")
```

```python
print("Missing CouponCode:", df["CouponCode"].isnull().sum())
```

```python
duplicate_orderids = df["OrderID"].duplicated().sum()

print("Duplicate OrderIDs:", duplicate_orderids)
```

```python
print("Date data type:", df["Date"].dtype)
print("Invalid dates:", pd.to_datetime(df["Date"], errors="coerce").isna().sum())
```

```python
# Convert Date column to proper date format
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# Format as YYYY-MM-DD
df["Date"] = df["Date"].dt.strftime("%Y-%m-%d")

# Check first 5 dates
print(df["Date"].head())
```

```python
duplicate_rows = df.duplicated().sum()

print("Duplicate complete rows:", duplicate_rows)
```

```python
print(df["Date"].head())
print("\nData type:", df["Date"].dtype)
```

```python
df["Date"] = pd.to_datetime(df["Date"]).dt.strftime("%Y-%m-%d")
```

```python
print(df["Date"].head())
```

```python
print("Quantity:", df["Quantity"].dtype)
print("ItemsInCart:", df["ItemsInCart"].dtype)
print("UnitPrice:", df["UnitPrice"].dtype)
print("TotalPrice:", df["TotalPrice"].dtype)
```

```python
df["Quantity"] = df["Quantity"].round().astype(int)
df["ItemsInCart"] = df["ItemsInCart"].round().astype(int)
```

```python
print(df[["Quantity", "ItemsInCart"]].head())
```

```python
df["UnitPrice"] = df["UnitPrice"].round(2)
df["TotalPrice"] = df["TotalPrice"].round(2)
```

```python
print(df[["UnitPrice", "TotalPrice"]].head())
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("\nBlank values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
df.to_excel("Project1_Cleaned.xlsx", index=False)

print("File saved successfully!")
```

```python
# Clean text columns
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.strip()

print("Text columns cleaned:")
print(list(text_columns))
```

```python
# Check numeric columns
numeric_columns = df.select_dtypes(include="number").columns

print("Numeric columns:")
print(list(numeric_columns))

print("\nData types:")
print(df[numeric_columns].dtypes)
```

```python
# Format numeric columns to 2 decimal places
for col in df.select_dtypes(include="number").columns:
    df[col] = df[col].round(2)

print("Numeric values formatted to 2 decimal places.")
```

```python
# Final quality checks

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nTotal blank values:", df.isnull().sum().sum())

print("Duplicate Order IDs:", df["OrderID"].duplicated().sum())

print("Invalid dates:", pd.to_datetime(df["Date"], errors="coerce").isna().sum())

print("\nFinal data types:")
print(df.dtypes)
```

```python
# Save final cleaned dataset
output_file = "DecodeLabs_Project1_Cleaned.xlsx"

df.to_excel(output_file, index=False)

print("Cleaned file saved as:", output_file)
```
